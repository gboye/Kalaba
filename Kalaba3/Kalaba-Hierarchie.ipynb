```python
class Hierarchie:
    def __init__(self):
        self.children={}
        self.parents={}

    def addChild(self,child,parent=None):
        if parent:
            if parent not in self.children:
                self.children[parent]=[]
            self.children[parent]+=[child]
            if not isinstance(child,list):
                self.parents[child]=[parent]+self.parents[parent]
        else:
            self.parents[child]=[]

    def getLeaves(self):
        leaves=[]
        for parent,children in self.children.items():
            for child in children:
                if child not in self.children:
                    leaves.append(child)
        return leaves
            
```

```python
cf=Hierarchie()
cf.addChild("N1")
cf.addChild("N11","N1")
cf.addChild("N111","N11")
cf.addChild("N112","N11")
cf.addChild("N12","N1")
cf.addChild("N2")
cf.addChild("N21","N2")
cf.addChild("N22","N2")
for leaf in cf.getLeaves():
    print(leaf,cf.parents[leaf])
cf.parents["N111"]
```

```python
class Lexique:
    def __init__(self):
        self.category={}
        self.stem={}
        self.french={}
        self.kalaba={}

    def addCategory(self,parent,child):
        # print("cat",parent,child)
        if parent in self.category:
            self.category[child]=[parent]+self.category[parent]
        else:
            self.category[child]=[parent]

    def addStem(self,stem,french):
        # print("stem",stem,french)        
        self.stem[stem]=self.category[stem]
        self.french[stem]=french
        for f in french:
            self.kalaba[f]=stem
    

lexique=Lexique()

def addLevel(l,level):
    for k,v in level.items():
        if isinstance(v,dict):
            for vv in v:
                l.addCategory(k,vv)
            addLevel(l,v)
        elif isinstance(v,list):
            l.addStem(k,v)

addLevel(lexique,stems["NOM"])
            
```
