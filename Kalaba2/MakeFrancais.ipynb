```python
import re
debug=False
```

```python
# -*- coding: utf8 -*-
from os.path import expanduser
home = expanduser("~")
annee="24"
nKalaba="K1"
repertoire=home+"/sDrive/Cours/Bordeaux/L1-LinguistiqueGenerale/00-ProjetKalaba/"+annee+"-%s"%nKalaba
serie=repertoire+"/"
```

```python
fFrancais=serie+"Phrases.csv"
fFrancaisCSV=serie+"Francais.txt"
with open(fFrancais,"r") as inFile:
    phrases=[l.strip() for l in inFile.readlines() if l[0]!="#"]
```

```python
fonctions="SUJ VER OBJ IND COMP AJOUT".split(" ")
```

```python
def cleanSyntagme(inS):
    if inS[0]==",":
        s=inS[1:]+inS[0]
    else:
        s=inS
    syntagme=[]
    mots=[m for m in s.split(" ") if m!=""]
    for mot in mots:
        clean=mot
        m=re.match(ur"^(\w+)#.*#$",clean)
        if m:
            clean=m.group(1)
        m=re.match(ur"^#(.+)#$",clean)
        if m:
            clean=m.group(1)
        if clean[-1] in "PD":
            clean=clean[:-1]
        
        clean=clean[0]+clean[1:].lower()
        syntagme.append(clean)
    return " ".join(syntagme)
```

```python
csvLines=[]
newPhrases=[]
newTagged=[]
for p in phrases:
    syntagmes=p.split("\t")
    if debug: print p
    newP=[]
    newT=[]
    for n,s in enumerate(syntagmes):
        if s!="":
            if ";" in s:
                sParts=s.split(";")
                for sPart in sParts:
                    if sPart[0]==",":
                        if debug: print "-1",fonctions[n],cleanSyntagme(sPart)
                        newP.insert(0,cleanSyntagme(sPart))
                        newT.insert(0,cleanSyntagme(sPart))
                        newT.insert(0,fonctions[n])
                    else:
                        if debug: print n,fonctions[n],cleanSyntagme(sPart)
                        newP.append(cleanSyntagme(sPart))
                        newT.append(fonctions[n])
                        newT.append(cleanSyntagme(sPart))
            else:
                if s[0]==",":
                    if debug: print "-1",fonctions[n],cleanSyntagme(s)
                    newP.insert(0,cleanSyntagme(s))
                    newT.insert(0,cleanSyntagme(s))
                    newT.insert(0,fonctions[n])
                else:
                    if debug: print n,fonctions[n],cleanSyntagme(s)
                    newP.append(cleanSyntagme(s))
                    newT.append(fonctions[n])
                    newT.append(cleanSyntagme(s))
    if debug: print
    newP[0]=newP[0].capitalize()
    newT[1]=newT[1].capitalize()
    newPhrases.append(" ".join(newP)+".")
    newTagged.append(" ".join(newT)+".")
    csvLines.append(" ".join(newP)+"."+";"+" ".join(newT)+".")
csvLines
```

```python
with open(fFrancaisCSV,"w") as outFile:
    for l in csvLines:
        outFile.write(l+"\n")
```
