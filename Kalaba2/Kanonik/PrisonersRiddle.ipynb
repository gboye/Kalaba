```python
import random as rd
nPrisoners=100
```

```python
def getBox(n):
    return box[n-1]

def getCycle(cycle):
    global fail
    nextN=getBox(cycle[-1])
    # print (nextN)
    cycle.append(nextN)
    if nextN!=cycle[0]:
        getCycle(cycle)
    else:
        # print (len(cycle))
        if len(cycle)>nPrisoners/2:
            fail=True
        return len(cycle)-1
```

```python
trials=10000
success=0
fails=0
for t in range(trials):
    # print (t,end=": ")
    fail=False
    box=list(range(1,nPrisoners))
    rd.shuffle(box)
    for i in range(1,nPrisoners):
        cycle=getCycle([i])
    if fail==True:
        fails+=1
        # print ("Failure")
    else:
        # print ("Success")
        success+=1
float(success)/trials
```
