```python
home = expanduser("~")
repertoire=home+"/ownCloud/Cours/Bordeaux/L1-UE4-Morphologie/00-Kanonik/"
serie=repertoire+"24-Kanoniks/"
serie=serie+"Kanonik-%02d/"%numerosKalaba[0]
print(serie)
```

```python
fichiers=glob.glob(serie+"*.*")
for n,fichier in enumerate(fichiers):
    print(n,fichier.split("/")[-1])
```

```python
fCopies={"Stems.yaml":7,"Phrases.csv":5,"Blocks.yaml":1,"Kanonik.csv":3}
```

```python
translit={"ʃ":"S","ŋ":"N","ʒ":"Z","θ":"T","ð":"D"}
```

```python
def IPA2Sampa(IPAlines):
    SampaLines=[]
    for IPAline in IPAlines:
        line=IPAline
        for l in translit:
            line=line.replace(l,translit[l])
        SampaLines.append(line)
    return SampaLines
```

```python
def TAB2Space(lines):
    result=[]
    for line in lines:
        result.append(line.replace("\t"," "))
    return result
```

```python
for fCopie,n in fCopies.items():
    with open(fichiers[n], 'r') as inFile:
        fLines=inFile.readlines()
    
    if fCopie.endswith(".yaml"):
        fLines=TAB2Space(fLines)
        fLines=IPA2Sampa(fLines)
       
    with open(serie+fCopie,'w') as outFile:
        for line in fLines:
            outFile.write(line)  
```

```python
nPhrases=fCopies["Phrases.csv"]
with open(fichiers[nPhrases], 'r') as inFile:
    fLines=inFile.readlines()

with open(serie+"Traductions.csv",'w') as outFile:
    for line in fLines[:5]:
        outFile.write(line)
```
