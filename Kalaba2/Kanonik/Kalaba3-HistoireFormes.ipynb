```python
# -*- coding: utf8 -*-
#########################IMPORTS############################################
from os.path import expanduser
import codecs
import re
import sys,os,time
import string
import yaml, YamlDuplicates
import pandas as pd
```

```python
def getTranslation(stemStructure):
    formes={}
    for element in stemStructure:
        if type(stemStructure[element])==list:
            listeFormes=stemStructure[element]
            lemme=listeFormes[0]
            formes[lemme]=element
        else:
            formes.update(getTranslation(stemStructure[element]))
    return formes
```

```python
debug=False
home = expanduser("~")
rep=home+"/sDrive/Cours/Bordeaux/L1-LinguistiqueGenerale/00-ProjetKalaba/"
annee=25

stems={}
for i in range(1,6):
    print i
    with open(rep+"%d-K%d/Stems.yaml"%(annee,i), 'r') as stream:
        stems[i]=getTranslation(yaml.load(stream,Loader=yaml.Loader))
# stems
```
