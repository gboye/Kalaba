```python
traduction_nom=serie+"Traductions.txt"
```

```python
dfTrads=pd.read_csv(traduction_nom,sep=";",header=None)
if nKalaba in [3,5]:
    dfTrads.columns=["klb","francais","kalaba"]
else:
    dfTrads.columns=["kalaba","francais"]
q1,q2,q3=12,17,22
dfTrads["l"]=dfTrads.francais.str.split(" ").str.len()
dfTrads[dfTrads.l<=q1].groupby("l").count()[["francais"]].plot(kind="bar",figsize=(20,5))
```

```python
dfTrads[(dfTrads.l>q1)&(dfTrads.l<q2)].groupby("l").count()[["francais"]].plot(kind="bar",figsize=(20,5))
```

```python
dfTrads[(dfTrads.l>q2)&(dfTrads.l<=q3)].groupby("l").count()[["francais"]].plot(kind="bar",figsize=(20,5))
```

```python
dfTrads[(dfTrads.l>q3)].groupby("l").count()[["francais"]].plot(kind="bar",figsize=(20,5))
```

```python
dfTrads.loc[dfTrads.l<q1].to_csv(traduction_nom.replace(".txt","-Q1.txt"),sep=";",header=False,index=False)
dfTrads.loc[(dfTrads.l>=q1)&(dfTrads.l<q2)].to_csv(traduction_nom.replace(".txt","-Q2.txt"),sep=";",header=False,index=False)
dfTrads.loc[(dfTrads.l>=q2)&(dfTrads.l<q3)].to_csv(traduction_nom.replace(".txt","-Q3.txt"),sep=";",header=False,index=False)
dfTrads.loc[dfTrads.l>=24].to_csv(traduction_nom.replace(".txt","-Q4.txt"),sep=";",header=False,index=False)
```
