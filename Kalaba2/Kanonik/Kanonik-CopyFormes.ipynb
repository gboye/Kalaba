```python
def getFrench(stemStructure):
    formes={}
    for element in stemStructure:
        if type(stemStructure[element])==list:
            listeFormes=stemStructure[element]
            lemme=listeFormes[0]
            formes[lemme]=listeFormes
        else:
            formes.update(getFrench(stemStructure[element]))
    return formes
```

```python
dPrev=getFrench(prevStems)
dNext=getFrench(nextStems)
```
