```python
import yaml,re,random
```

```python
repBase="/Volumes/BroadExt/Dropbox/Partage-Broadway/GAM/Caches/NosCaches/Kalabas/Bayonne/"
annee="19"
kalaba="K1"
rep=repBase+annee+"-"+kalaba+"/"
```

```python
with open(rep+"phonology.yaml", 'r') as stream:
    phonology=yaml.safe_load(stream)
```

```python
monogrammes=phonology["translations"]["monogrammes"]
graphoIn=phonology["translations"]["grapho"]["in"]
graphoOut=phonology["translations"]["grapho"]["out"]
translit = dict(zip(graphoIn, graphoOut))
translit
```

```python
graphies={k:(monogrammes[v] if v in monogrammes else v) for k,v in translit.items()}
if kalaba=="K1":
    print("écriture spéciale pour Z=qe et v=Ra")
    graphies["Z"]="qe"
    graphies["v"]="RI"
```

```python
graphies
```

```python
alphabetRaw=[]
for k,v in graphies.items():
    alphabetRaw.append("%s: \\begin{persian}\\sylla{%s}\\end{persian}\\\\"%(k,v))
random.shuffle(alphabetRaw)
```

```python
alphabetKey=[]
alphabetDict={}
for i,l in enumerate(alphabetRaw):
    m=re.match("^(\w): (.*)\\\\$",l)
    if m:
        k=m.group(1)
        v=m.group(2)
    alphabetLine="%02d: "%(i+1)+l
    alphabetDict[k]=alphabetLine.strip("\\")
    alphabetKey.append(alphabetLine)
print ("\n".join(alphabetKey))
print(alphabetDict)
```

```python
alphabetDict["v"]='20: v: \\begin{persian}\\sylla{RI}\\end{persian}'
alphabetKey[19]='20: v: \\begin{persian}\\sylla{RI}\\end{persian}\\\\'
alphabetKey.append('27: v: \\begin{persian}\\sylla{Ru}\\end{persian}\\\\')
alphabetKey.append('28: Ž: \\begin{persian}\\sylla{q}\\end{persian}\\\\')
```

```python
def chunker(seq, size):
    for pos in range(0, len(seq), size): 
        yield seq[pos:pos + size]
```

```python
alphabet=[]
for l in alphabetKey:
    m=re.match(r"(\d\d: )\w: (.*)\\\\",l)
    if m:
        alphabet.append(m.group(1)+m.group(2))
print ("\n".join(alphabet))
```

```python
print("\\begin{tabular}{lllll}\\toprule")
for c in chunker(alphabet,5):
    print("&".join(c)+"\\\\")
print("\\bottomrule\\end{tabular}")
```

```python
print("\\begin{tabular}{llllll}\\toprule")
for c in chunker(graphoIn,6):
    lineKey=[alphabetDict[k] for k in c]
    print("&".join(lineKey)+"\\\\")
print("\\bottomrule\\end{tabular}")
```

```python
numAlphabet={}
for l in alphabetKey:
    m=re.match("(\d\d):\s*(\w):.*persian.(.*)\\\\end",l)
    if m:
        num=m.group(1)
        lettre=m.group(2)
        ecrit=m.group(3)
        numAlphabet[num]=ecrit
```
