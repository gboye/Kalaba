```python
import re
```

```python
repBase="/Volumes/BroadExt/Dropbox/Partage-Broadway/GAM/Caches/NosCaches/Kalabas/Bayonne/"
annee="19"
kalaba="K2"
rep=repBase+annee+"-"+kalaba+"/"
```

## Adaptation des Kalabas avec espaces

### Exemples-expex

```python
nExpex="Exemples-expex.tex"
with open(rep+nExpex) as inFile:
    lines=[l.strip() for l in inFile.readlines()]
```

```python
newLines=[]
for line in lines:
    m=re.match(r"^(\\gl[ab]\s*|% \\glc\s+)(.*)//$",line)
    if m:
        exType=m.group(1)
        mots=[mot for mot in m.group(2).split("{}") if mot.strip()!=""]  # éliminer l'espace à la fin
        mots.reverse()
        # print(mots)
        newLines.append(exType+"{}".join(mots)+"{}//")
    else:
        newLines.append(line)
```

```python
with open(rep+nExpex.replace(".tex","-RL.tex"),"w") as outFile:
    for line in newLines:
        outFile.write(line+"\n")
```

### Adaptation des Images

```python
nImages="Images.tex"
with open(rep+nImages) as inFile:
    lines=[l.strip() for l in inFile.readlines()]
```

```python
newLines=[]
nEx=0
bExpex="\\begin{preview}\n"
eExpex="\n\\end{preview}\n"
for line in lines:
    m=re.match(r"^(\\begin{preview})(.*)(\\end{preview})\s*$",line)
    mm=re.match(r"^%(\\begin{preview})(.*)(\\end{preview})\s*$",line)
    mmm=re.match(r"^%\s*(.*)\\end{preview}\s*$",line)
    if m:
        mots=m.group(2).split(" ")
        mots.reverse()
        if nEx==0:
            newLines.append(bExpex+" ".join(mots)+eExpex)
        else:
            newLines.append(bExpex+"\\begin{center}"+" ".join(mots)+"\\\\\\begin{french}(%s%s-%d)\\end{french}\\end{center}"%(annee,kalaba[-1],nEx)+eExpex)
        nEx+=1
    elif mm:
        newLines.append(bExpex+"\\begin{french}(%s%s-%d) "%(annee,kalaba[-1],nEx-1)+mm.group(2)+"\\end{french}"+eExpex)
    elif mmm:
        glose=[]
        mots=mmm.group(1).split(" ")
        mots.reverse()
        for mot in mots:
            glose.append(mot.replace("P{}","G{}"))
        newLines.append(bExpex+"\\begin{center}\n\\begin{french}\n"+" ".join(mots)+"\\\\")
        newLines.append(" ".join(glose)+"\\\\\n\\end{french}\n\\end{center}"+eExpex)           
    else:
        newLines.append(line)
```

```python
with open(rep+nImages.replace(".tex","-RL.tex"),"w") as outFile:
    for line in newLines:
        outFile.write(line+"\n")
```

## Adaptation des Kalabas sans espaces

### Exemples

```python
import re
```

```python
repBase="/Volumes/BroadExt/Dropbox/Partage-Broadway/GAM/Caches/NosCaches/Kalabas/Bayonne/"
annee="19"
kalaba="K3"
rep=repBase+annee+"-"+kalaba+"/"
```

```python
nExpex="Exemples-expex.tex"
with open(rep+nExpex) as inFile:
    lines=[l.strip() for l in inFile.readlines()]
```

```python
newLines=[]
exCounter=1
for line in lines:
    vline=line
    vm=re.search(r"Trois(Sg|Du|Pl)(.)(Sg|Du|Pl)(P?\{\})",vline)
    if vm:
        vline=vline.replace(vm.group(0),"Trois"+vm.group(1)+vm.group(2)+vm.group(4))
    m=re.match(r"^(\\gl[ab]\s*|\\glft\s+)(.*)//$",vline)
    if m:
        exType=m.group(1)
        # print(exType)
        exContent=m.group(2)
        if "gla" in exType:
            newLines.append("\\raggedleft\\needspace{4\\baselineskip}\n")
            newLines.append(exContent+" (%02d)\\\\\n"%exCounter)
            exCounter+=1
        elif "glb" in exType:
            newLines.append(exContent+" \\textcolor{white}{(%02d)}\\\\\n"%exCounter)
        elif "glft" in exType:
            # newLines.append("\\raggedright\n"+exContent+"\\\\\n\\bigskip\n\n")
            newLines.append(exContent+" \\textcolor{white}{(%02d)}\\\\\n\\bigskip\n\n"%exCounter)
    else:
        newLines.append("% "+line)
```

```python
with open(rep+nExpex.replace(".tex","-GC.tex"),"w") as outFile:
    for line in newLines:
        outFile.write(line+"\n")
```

### Adaptation des Images

```python
nImages="Images.tex"
with open(rep+nImages) as inFile:
    lines=[l.strip() for l in inFile.readlines()]
```

```python
newLines=[]
nEx=0
bExpex="\\begin{preview}\n"
eExpex="\n\\end{preview}\n"
for line in lines:
    m=re.match(r"^(\\begin{preview})(.*)(\\end{preview})\s*$",line)
    mm=re.match(r"^%(\\begin{preview})(.*)(\\end{preview})\s*$",line)
    mmm=re.match(r"^%\s*(.*)\\end{preview}\s*$",line)
    if m:
        ecriture=m.group(2)
        newLines.append(bExpex+ecriture+" (%s%s-%d)"%(annee,kalaba[-1],nEx)+eExpex)
        nEx+=1
    elif mm:
        newLines.append(bExpex+"(%s%s-%d) "%(annee,kalaba[-1],nEx-1)+mm.group(2)+eExpex)
    elif mmm:
        glose=[]
        decoupe=[]
        phonetique=[]
        mots=mmm.group(1).split("{}")
        mots.reverse()
        for mot in mots[1:]:
            phonetique.append(mot+"{}")
            decoupe.append(mot[:-1]+"D{} ")
            glose.append(mot[:-1]+"G{} ")
        newLines.append(bExpex+"\\begin{center}\n"+"".join(phonetique)+"\n\\end{center}\n"+eExpex)
        newLines.append(bExpex+"\\begin{center}\n"+"".join(decoupe)+"\n\\end{center}\n"+eExpex)
        newLines.append(bExpex+"\\begin{center}\n"+"".join(glose)+"\n\\end{center}\n"+eExpex)
    else:
        newLines.append(line)
```

```python
with open(rep+nImages.replace(".tex","-GC.tex"),"w") as outFile:
    for line in newLines:
        outFile.write(line+"\n")
```
